Extracts Grad-CAM activation maps from the classification backbone and renders the 5 interactive diagnostic overlays (`original`, `gradcam`, `detection`, `segmentation`, `combined`).


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path('..').resolve()))

from src.classification.inference import ClassificationInferenceService
from src.explainability.gradcam import GradCAMExtractor
from src.data.preprocessing import preprocess_xray
from pathlib import Path

sample_bmp = next(Path('../dataset/raw/Segmentation/test').glob('*.bmp'))
cls_service = ClassificationInferenceService('..')
cam = GradCAMExtractor(cls_service.model, cls_service.model.get_target_conv_layer())
prep = preprocess_xray(sample_bmp, target_size=cls_service.image_size, three_channel=True)
heatmap = cam.generate_heatmap(prep['tensor'].unsqueeze(0), output_size=(512, 512))
print('Generated Grad-CAM heatmap shape:', heatmap.shape, 'min/max:', heatmap.min(), heatmap.max())
```
